```python
#(i)
import random
random.seed(1024160003)

numbers = [random.randint(100, 900) for _ in range(100)]
print("Random list: ", numbers)
```

```python
#(ii)
odd_numbers = [num for num in numbers if num % 2 != 0]
print("Number of Odd numbers: ", len(odd_numbers))
```

```python
#(iii)
even_numbers = [num for num in numbers if num % 2 == 0]
print("Number of Even numbers: ", len(even_numbers))
```

```python
#(iv)
def is_prime(n):
    if n <= 2:
      return False
    for i in range(2, int(n**0.5) + 1):
      if n % i == 0:
        return False
    return True
prime_numbers = [num for num in numbers if is_prime(num)]
print("Number of prime numbers: ", len(prime_numbers))
print("Prime Numbers: ", prime_numbers)
```

```python
#(v)
frequency = {}
for num in numbers:
  frequency[num] = frequency.get(num, 0) + 1
most_frequent = max(frequency, key = frequency.get)
print("Most frequent number: ", most_frequent)
print("frequency of most frequent: ", frequency[most_frequent])
```

```python
digits = [1, 0, 2, 4, 1, 6, 0, 0]
A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}
print("set A: ", A)
print("set B: ", B)
```

```python
#(i)
print("Union of A and B: ", A.union(B))
```

```python
#(ii)
print("Intersection of A and B: ", A.intersection(B))
```

```python
#(iii)
print("A - B: ", A.difference(B))
print("B - A: ", B.difference(A))
```

```python
#(iv)
print("Symmetric difference: ", A.symmetric_difference(B))
```

```python
#(v)
print("Is A a subset of B ?", A.issubset(B))
print("Is B a subset of A ?", B.issubset(A))
```

```python
#(vi)
X = int(input("Enter a value to remove from set A: "))
A.discard(X)
print("Updated set A: ", A)
```
