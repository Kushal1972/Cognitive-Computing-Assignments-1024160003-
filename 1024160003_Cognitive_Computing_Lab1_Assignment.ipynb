```python
#6.1
def AddOdd(n):
  sum = 0
  for s in range(1,n+1):
    if s % 2 != 0:
      sum = sum + s
  print(sum)
n = int(input("Enter the number: "))
AddOdd(n)
```

```python
#6.2
def AddPrime(n):
  total = 0
  for i in range(1,n+1):
    prime = True
    for j in range(2,i):
      if i % j == 0:
        prime = False
        break
    if prime:
      total = total + i
  print(total)
n = int(input("Enter the number: "))
AddPrime(n)
```
